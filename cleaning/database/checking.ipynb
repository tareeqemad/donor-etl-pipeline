# TESTING DATABASE OF THE SEVER

```python
import psycopg2
from configparser import ConfigParser

def read_db_config(filename='database_server.ini', section='postgresql'):
    parser = ConfigParser()
    parser.read(filename)

    # Get section, return as a dictionary
    db_config = {}
    if parser.has_section(section):
        params = parser.items(section)
        for param in params:
            db_config[param[0]] = param[1]
    else:
        raise Exception(f"Section {section} not found in {filename}")
    return db_config

# readiing config
db_params = read_db_config()

# Connect 
try:
    conn = psycopg2.connect(**db_params)
    print("Connected successfully!")
    cur = conn.cursor()
    cur.execute("SELECT * FROM public.contributors LIMIT 5;")
    rows = cur.fetchall()
    for row in rows:
        print(row)
    cur.close()
    conn.close()
except Exception as e:
    print("Error:", e)
```
